```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Recover original, unscaled training data
X_train_raw = df.loc[X_train.index].drop(columns=["Class"])
y_train_raw = df.loc[X_train.index, "Class"]

# Recover original, unscaled testing data
X_test_raw = df.loc[X_test.index].drop(columns=["Class"])

# Split training data into subtraining and validation
X_subtrain_raw, X_val_raw, y_subtrain, y_val = train_test_split(
    X_train_raw,
    y_train_raw,
    test_size=0.20,
    random_state=42,
    stratify=y_train_raw
)

# Create a fresh scaler
tuning_scaler = StandardScaler()

scale_columns = ["Time", "Amount", "Log_Amount"]

X_subtrain = X_subtrain_raw.copy()
X_val = X_val_raw.copy()
X_test_final = X_test_raw.copy()

# Fit only on subtraining data
X_subtrain[scale_columns] = tuning_scaler.fit_transform(
    X_subtrain[scale_columns]
)

# Transform validation and test data
X_val[scale_columns] = tuning_scaler.transform(
    X_val[scale_columns]
)

X_test_final[scale_columns] = tuning_scaler.transform(
    X_test_final[scale_columns]
)

print("Subtraining Shape:", X_subtrain.shape)
print("Validation Shape:", X_val.shape)
print("Testing Shape:", X_test_final.shape)

print("\nSubtraining Distribution:")
print(y_subtrain.value_counts())

print("\nValidation Distribution:")
print(y_val.value_counts())
```

```python
from imblearn.over_sampling import SMOTE

smote_tuning = SMOTE(
    random_state=42,
    sampling_strategy="auto"
)

X_subtrain_smote, y_subtrain_smote = smote_tuning.fit_resample(
    X_subtrain,
    y_subtrain
)
```

```python
print("Before SMOTE:")
print(y_subtrain.value_counts())

print("\nAfter SMOTE:")
print(y_subtrain_smote.value_counts())

print("\nNew Training Shape:", X_subtrain_smote.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
```

```python
rf_pipeline = Pipeline([
    ("smote", SMOTE(random_state=42)),
    ("classifier", RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
param_grid = {
    "classifier__n_estimators": [100, 150, 200],
    "classifier__max_depth": [10, 15, 20, None],
    "classifier__min_samples_split": [2, 5, 10],
    "classifier__min_samples_leaf": [1, 2, 4],
    "classifier__max_features": ["sqrt", "log2"]
}
```

```python
random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_grid,
    n_iter=5,
    scoring="f1",
    cv=3,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

random_search.fit(X_subtrain, y_subtrain)
```

```python
print("Best Parameters:")
print(random_search.best_params_)

print("\nBest Cross Validation F1 Score:")
print(random_search.best_score_)
```

```python
print("Best Parameters:")

for parameter, value in random_search.best_params_.items():
    print(f"{parameter}: {value}")
```

```python
best_rf = random_search.best_estimator_

y_val_proba = best_rf.predict_proba(X_val)[:, 1]

print("Validation Predictions Generated!")
print("Total Validation Records:", len(y_val_proba))
```

```python
from sklearn.metrics import precision_recall_curve
import numpy as np

precision, recall, thresholds = precision_recall_curve(
    y_val,
    y_val_proba
)

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

best_index = np.argmax(f1_scores)

best_threshold = thresholds[best_index]

print("Best Threshold:", best_threshold)
print("Best Validation Precision:", precision[best_index])
print("Best Validation Recall:", recall[best_index])
print("Best Validation F1 Score:", f1_scores[best_index])
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

y_val_pred = (y_val_proba >= best_threshold).astype(int)

print(classification_report(y_val, y_val_pred))

cm = confusion_matrix(y_val, y_val_pred)

plt.figure(figsize=(7, 5))

sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")

plt.title("Optimized Random Forest - Validation")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
print("Best Threshold:", best_threshold)
print("Validation Precision:", precision[best_index])
print("Validation Recall:", recall[best_index])
print("Validation F1:", f1_scores[best_index])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    classification_report,
    confusion_matrix
)

# Predict fraud probabilities
y_test_proba = best_rf.predict_proba(X_test_final)[:, 1]

# Apply selected validation threshold
y_test_optimized = (
    y_test_proba >= best_threshold
).astype(int)

# Evaluation
print("FINAL OPTIMIZED RANDOM FOREST RESULTS")

print("Accuracy:", accuracy_score(y_test, y_test_optimized))
print("Precision:", precision_score(y_test, y_test_optimized))
print("Recall:", recall_score(y_test, y_test_optimized))
print("F1 Score:", f1_score(y_test, y_test_optimized))

print("PR-AUC:", average_precision_score(y_test, y_test_proba))

print("\nClassification Report:")
print(classification_report(y_test, y_test_optimized))
```

```python
cm_final = confusion_matrix(y_test, y_test_optimized)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_final,
    annot=True,
    fmt="d",
    cmap="Greens"
)

plt.title("Final Optimized Random Forest")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```
